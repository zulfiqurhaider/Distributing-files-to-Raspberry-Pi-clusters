```python
import paramiko
import pandas as pd
from datetime import datetime
start_time = datetime.now()
# Load the dataset
dataset_path = "/Users/mhm36021/Downloads/archive/cleaned_reviews.csv"
df = pd.read_csv(dataset_path)

# Split the dataset into three portions, making sure each portion contains the header
num_rows = len(df)
part_size = num_rows // 3

for i in range(3):
    start = i * part_size
    end = (i + 1) * part_size if i < 2 else num_rows

    # Create a new DataFrame containing the subset of the data
    part_df = df[start:end]

    # Save each portion to a separate CSV file with the header
    part_filename = f"portion_{i + 1}.csv"
    part_df.to_csv(part_filename, index=False)

print("Dataset split into three portions, each containing the header.")

# Define the SSH parameters for the Raspberry Pis
pi1_ip = '192.168.1.39'
pi2_ip = '192.168.1.40'
pi3_ip = '192.168.1.41'
pi_username = 'admin'
pi_password = 'password'

# Define the destination paths on the Raspberry Pis
pi1_destination_path = '/home/admin/'
pi2_destination_path = '/home/admin/'
pi3_destination_path = '/home/admin/'

# Create SSH clients for each Raspberry Pi
ssh1 = paramiko.SSHClient()
ssh1.set_missing_host_key_policy(paramiko.AutoAddPolicy())

ssh2 = paramiko.SSHClient()
ssh2.set_missing_host_key_policy(paramiko.AutoAddPolicy())

ssh3 = paramiko.SSHClient()
ssh3.set_missing_host_key_policy(paramiko.AutoAddPolicy())

try:
    # Connect to each Raspberry Pi
    ssh1.connect(pi1_ip, username=pi_username, password=pi_password)
    ssh2.connect(pi2_ip, username=pi_username, password=pi_password)
    ssh3.connect(pi3_ip, username=pi_username, password=pi_password)

    # Use SFTP to send each portion to the respective Raspberry Pi
    sftp1 = ssh1.open_sftp()
    sftp2 = ssh2.open_sftp()
    sftp3 = ssh3.open_sftp()

    for i, (remote_file, sftp) in enumerate(
        [(pi1_destination_path, sftp1), (pi2_destination_path, sftp2), (pi3_destination_path, sftp3)]
    ):
        part_filename = f"portion_{i + 1}.csv"
        local_part_path = part_filename
        remote_part_path = remote_file + part_filename
        sftp.put(local_part_path, remote_part_path)

    print('Successfully sent portions of the dataset to the Raspberry Pis.')

except Exception as e:
    print(f'Error: {e}')

finally:
    # Close the SSH connections
    ssh1.close()
    ssh2.close()
    ssh3.close()
end_time = datetime.now()
print('Duration: {}'.format(end_time - start_time))
```

```python
from datetime import datetime
start_time = datetime.now()
def command_raspberry_pis_to_execute_script():
    # Define the SSH parameters for the Raspberry Pis
    pi1_ip = '192.168.1.39'
    pi2_ip = '192.168.1.40'
    pi3_ip = '192.168.1.41'
    pi_username = 'admin'
    pi_password = 'password'

    # Define the script file path on the Raspberry Pis
    script_path = '/home/admin/Full.py'  # Replace with the actual script file path on each Raspberry Pi

    # Define the command to execute the script (modify as needed)
    command = f'python3 {script_path}'

    # Create SSH clients for each Raspberry Pi
    ssh1 = paramiko.SSHClient()
    ssh1.set_missing_host_key_policy(paramiko.AutoAddPolicy())

    ssh2 = paramiko.SSHClient()
    ssh2.set_missing_host_key_policy(paramiko.AutoAddPolicy())

    ssh3 = paramiko.SSHClient()
    ssh3.set_missing_host_key_policy(paramiko.AutoAddPolicy())

    try:
        # Connect to each Raspberry Pi
        ssh1.connect(pi1_ip, username=pi_username, password=pi_password)
        ssh2.connect(pi2_ip, username=pi_username, password=pi_password)
        ssh3.connect(pi3_ip, username=pi_username, password=pi_password)

        # Execute the script remotely on each Raspberry Pi
        _, stdout1, stderr1 = ssh1.exec_command(command)
        _, stdout2, stderr2 = ssh2.exec_command(command)
        _, stdout3, stderr3 = ssh3.exec_command(command)

        # Print the output and errors (if any) from each Raspberry Pi
        print(f'Raspberry Pi 1 Output: {stdout1.read().decode()}')
        print(f'Raspberry Pi 1 Errors: {stderr1.read().decode()}')
        print(f'Raspberry Pi 2 Output: {stdout2.read().decode()}')
        print(f'Raspberry Pi 2 Errors: {stderr2.read().decode()}')
        print(f'Raspberry Pi 3 Output: {stdout3.read().decode()}')
        print(f'Raspberry Pi 3 Errors: {stderr3.read().decode()}')

        print('Successfully sent commands to execute the script on the Raspberry Pis.')

    except Exception as e:
        print(f'Error: {e}')

    finally:
        # Close the SSH connections
        ssh1.close()
        ssh2.close()
        ssh3.close()

if __name__ == '__main__':
    command_raspberry_pis_to_execute_script()
    
end_time = datetime.now()
print('Duration: {}'.format(end_time - start_time))
```
